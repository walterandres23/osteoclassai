# OsteoClassAI v2 · 14 Calibración local: ¿cuántas rodillas hacen falta?

En otro hospital los modelos conservan el orden de los grados pero gradúan más alto. La calibración local ajusta los cortes del grado esperado con unas pocas rodillas graduadas por los expertos del sitio nuevo.

Usa las predicciones ya guardadas de Gornale con el **recorte del detector** (cuaderno 12), así que no necesita GPU. Para cada tamaño de muestra (10 a 320 rodillas) toma 200 muestras al azar, ajusta con ellas y mide en el resto de rodillas del sitio. Compara:
- sin calibrar (argmax, como en la tabla del README);
- **desplazamiento** de los cortes (1 parámetro);
- **4 cortes** libres.

Exploratorio: Gornale ya se usó para decidir.

*Entorno de ejecución:* CPU. Luego *Ejecutar todas* (~2 min).
